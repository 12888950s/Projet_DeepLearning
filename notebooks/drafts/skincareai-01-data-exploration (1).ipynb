# SkinCare AI — Data Exploration

## Objective

This notebook explores the skin disease image dataset before building the deep learning models.

The main objectives are:
- Load the dataset
- Display available classes
- Select the target classes
- Count images per class
- Visualize sample images
- Analyze class distribution
- Justify the exclusion of noisy classes

This step helps understand the dataset before training CNN-based image classification models.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
```

```python
import os

DATASET_PATH = "/kaggle/input/datasets/pacificrm/skindiseasedataset/SkinDisease/SkinDisease"

TRAIN_DIR = os.path.join(DATASET_PATH, "train")
TEST_DIR = os.path.join(DATASET_PATH, "test")

print("Train exists:", os.path.exists(TRAIN_DIR))
print("Test exists:", os.path.exists(TEST_DIR))
```

## Dataset Structure

The dataset is organized into two main folders: `train` and `test`.

Each subfolder represents a skin disease category.  
The original dataset contains several dermatological classes, but only a subset of four classes will be used in this project to build a focused and controlled classification system.

```python
train_classes = sorted(os.listdir(TRAIN_DIR))
test_classes = sorted(os.listdir(TEST_DIR))

print("Number of classes in train:", len(train_classes))
print(train_classes)

print("\nNumber of classes in test:", len(test_classes))
print(test_classes)
```

## Selected Classes

After visual inspection, the `Unknown_Normal` class was excluded because it contains several non-skin images such as objects, rooms and food.  
Using this class as a normal skin category could mislead the model and introduce noise into the training process.

For this reason, the final classification task focuses on four dermatological disease classes:

- Acne
- Eczema
- Psoriasis
- Rosacea

```python
SELECTED_CLASSES = [
    "Acne",
    "Eczema",
    "Psoriasis",
    "Rosacea"
]

display_names = {
    "Acne": "Acne",
    "Eczema": "Eczema",
    "Psoriasis": "Psoriasis",
    "Rosacea": "Rosacea"
}

print("Selected classes:")
for cls in SELECTED_CLASSES:
    print("-", cls)
```

## Image Count per Class

Before training the models, the number of images per selected class is counted for both training and testing sets.

This step helps identify possible class imbalance, which may affect the model performance.

```python
def count_images(directory, selected_classes):
    data = []

    for cls in selected_classes:
        class_path = os.path.join(directory, cls)

        images = [
            img for img in os.listdir(class_path)
            if img.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".webp"))
        ]

        data.append({
            "Class": cls,
            "Number of Images": len(images)
        })

    return pd.DataFrame(data)

train_counts = count_images(TRAIN_DIR, SELECTED_CLASSES)
test_counts = count_images(TEST_DIR, SELECTED_CLASSES)

print("Training set:")
display(train_counts)

print("Testing set:")
display(test_counts)
```

```python
total_train = train_counts["Number of Images"].sum()
total_test = test_counts["Number of Images"].sum()

print("Total training images:", total_train)
print("Total testing images:", total_test)
print("Total selected images:", total_train + total_test)
```

## Class Distribution

The following bar charts show the distribution of images across the selected classes.

```python
plt.figure(figsize=(8, 5))
plt.bar(train_counts["Class"], train_counts["Number of Images"])
plt.title("Training Images per Class")
plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.bar(test_counts["Class"], test_counts["Number of Images"])
plt.title("Testing Images per Class")
plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.xticks(rotation=45)
plt.show()
```

## Sample Images

Sample images were inspected during the data exploration phase.  
For privacy and presentation reasons, direct image visualization was removed from the public notebook.

```python
def show_sample_images(directory, selected_classes, images_per_class=4):
    plt.figure(figsize=(14, 10))
    index = 1

    for cls in selected_classes:
        class_path = os.path.join(directory, cls)

        images = [
            img for img in os.listdir(class_path)
            if img.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".webp"))
        ]

        for img_name in images[:images_per_class]:
            img_path = os.path.join(class_path, img_name)
            img = Image.open(img_path).convert("RGB")

            plt.subplot(len(selected_classes), images_per_class, index)
            plt.imshow(img)
            plt.title(cls)
            plt.axis("off")
            index += 1

    plt.tight_layout()
    plt.show()

#show_sample_images(TRAIN_DIR, SELECTED_CLASSES, images_per_class=4)
```

## Dataset Analysis and Conclusion

The original dataset contains multiple skin disease classes.  
For this project, four classes were selected: Acne, Eczema, Psoriasis and Rosacea.

The `Unknown_Normal` class was excluded because visual inspection showed that it contains non-skin images. This decision helps reduce noise and improves the reliability of the classification task.

The class distribution was analyzed using image counts and bar charts. If class imbalance exists, it may affect the model by making it biased toward majority classes.
